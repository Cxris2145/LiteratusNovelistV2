```python
# ==============================================================
# PROYECTO: Generación masiva de portadas mediante IA
# Modelo: Stable Diffusion XL Turbo
# ==============================================================
#
# DESCRIPCIÓN
# -----------
# Este notebook genera automáticamente una portada para cada
# libro definido en el archivo "books_to_generate.json".
#
# Cada elemento del JSON debe contener:
#
# {
#     "slug": "nombre-archivo",
#     "title": "Título del libro",
#     "author": "Autor",
#     "prompt": "Descripción para generar la imagen"
# }
#
# El programa realiza las siguientes tareas:
#
# 1. Instala las dependencias necesarias.
# 2. Carga el archivo JSON.
# 3. Carga el modelo SDXL Turbo.
# 4. Genera una portada por cada libro.
# 5. Guarda todas las imágenes.
# 6. Comprime el resultado en un archivo ZIP.
#
# ==============================================================
# INSTRUCCIONES DE USO
# ==============================================================
#
# 1. Activar GPU en Google Colab:
#    Entorno de ejecución -> Cambiar tipo de entorno -> GPU
#
# 2. Subir el archivo:
#       books_to_generate.json
#
# 3. Ejecutar todas las celdas.
#
# 4. Descargar:
#       todas_las_portadas.zip
#
# ==============================================================

# --------------------------------------------------------------
# PASO 1 - Instalar dependencias
# --------------------------------------------------------------

!pip install -q diffusers transformers accelerate

# --------------------------------------------------------------
# PASO 2 - Importar librerías
# --------------------------------------------------------------

import json
import os
import time

import torch

from diffusers import StableDiffusionXLPipeline
from IPython.display import clear_output

# --------------------------------------------------------------
# PASO 3 - Configuración del proyecto
# --------------------------------------------------------------

JSON_FILE = "books_to_generate.json"

OUTPUT_FOLDER = "covers_finales"

MODEL_NAME = "stabilityai/sdxl-turbo"

NEGATIVE_PROMPT = (
    "text, letters, words, typography, blurry, low quality, "
    "deformed, ugly, watermark, signature, frames, borders"
)

NUM_INFERENCE_STEPS = 2

GUIDANCE_SCALE = 0.0

# --------------------------------------------------------------
# PASO 4 - Comprobar que existe el archivo JSON
# --------------------------------------------------------------

if not os.path.exists(JSON_FILE):
    raise FileNotFoundError(
        f"No se encontró el archivo '{JSON_FILE}'. "
        "Debe subirse antes de ejecutar el notebook."
    )

# --------------------------------------------------------------
# PASO 5 - Cargar el JSON
# --------------------------------------------------------------

with open(JSON_FILE, "r", encoding="utf-8") as f:
    tasks = json.load(f)

print(f"Se han encontrado {len(tasks)} libros.")

# --------------------------------------------------------------
# PASO 6 - Seleccionar dispositivo
# --------------------------------------------------------------

device = "cuda" if torch.cuda.is_available() else "cpu"

print(f"Dispositivo utilizado: {device}")

# --------------------------------------------------------------
# PASO 7 - Cargar el modelo
# --------------------------------------------------------------

pipe = StableDiffusionXLPipeline.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    variant="fp16" if device == "cuda" else None,
)

pipe.to(device)

# --------------------------------------------------------------
# PASO 8 - Crear carpeta de salida
# --------------------------------------------------------------

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# --------------------------------------------------------------
# PASO 9 - Generación de imágenes
# --------------------------------------------------------------

inicio = time.time()

imagenes_generadas = 0

imagenes_omitidas = 0

errores = []

print("\nIniciando generación...\n")

for i, task in enumerate(tasks, start=1):

    output_path = os.path.join(
        OUTPUT_FOLDER,
        f"{task['slug']}.jpg"
    )

    # Evitar volver a generar imágenes ya existentes
    if os.path.exists(output_path):
        imagenes_omitidas += 1
        continue

    try:

        if i == 1 or i % 10 == 0:

            clear_output(wait=True)

            porcentaje = (i / len(tasks)) * 100

            print("=" * 60)
            print("GENERACIÓN DE PORTADAS")
            print("=" * 60)
            print(f"Libro actual : {task['title']}")
            print(f"Progreso     : {i}/{len(tasks)}")
            print(f"Completado   : {porcentaje:.1f}%")
            print("=" * 60)

        image = pipe(
            prompt=task["prompt"],
            negative_prompt=NEGATIVE_PROMPT,
            num_inference_steps=NUM_INFERENCE_STEPS,
            guidance_scale=GUIDANCE_SCALE
        ).images[0]

        image.save(output_path)

        imagenes_generadas += 1

        # Liberar memoria de la GPU
        if device == "cuda":
            torch.cuda.empty_cache()

    except Exception as e:

        errores.append(
            {
                "titulo": task["title"],
                "error": str(e)
            }
        )

# --------------------------------------------------------------
# PASO 10 - Resumen
# --------------------------------------------------------------

tiempo = time.time() - inicio

print("\n")
print("=" * 60)
print("PROCESO FINALIZADO")
print("=" * 60)
print(f"Total de libros      : {len(tasks)}")
print(f"Imágenes generadas   : {imagenes_generadas}")
print(f"Imágenes omitidas    : {imagenes_omitidas}")
print(f"Errores encontrados  : {len(errores)}")
print(f"Tiempo total         : {tiempo/60:.2f} minutos")
print("=" * 60)

# --------------------------------------------------------------
# PASO 11 - Mostrar errores (si existen)
# --------------------------------------------------------------

if errores:

    print("\nListado de errores:\n")

    for error in errores:

        print(f"- {error['titulo']}")
        print(f"  {error['error']}\n")

# --------------------------------------------------------------
# PASO 12 - Comprimir resultados
# --------------------------------------------------------------

print("\nComprimiendo imágenes...")

!zip -rq todas_las_portadas.zip covers_finales

print("Archivo ZIP creado correctamente.")

print("\nProceso terminado con éxito.")
```
